# Employee Salary Analysis
**Track:** AI & ML 
**Objective:** Practice numerical data processing using Python to calculate total, average, highest, and lowest salaries.

```python
# 1. Manually create employee salary data
salaries = [45000, 52000, 61000, 48000, 75000, 38000, 59000, 82000, 41000, 67000]

# 2. Perform basic calculations using built-in functions
total_salary = sum(salaries)
num_employees = len(salaries)
average_salary = total_salary / num_employees
highest_salary = max(salaries)
lowest_salary = min(salaries)

# 3. Output the results
print("--- Standard Python Analysis ---")
print(f"Total Employees: {num_employees}")
print(f"Total Salary: ${total_salary:,}")
print(f"Average Salary: ${average_salary:,.2f}")
print(f"Highest Salary: ${highest_salary:,}")
print(f"Lowest Salary: ${lowest_salary:,}")
```

## Advanced Analysis using Pandas
As an extension of the basic requirements, the following section uses the `pandas` library to analyze the same dataset. This demonstrates a more industry-standard approach to data manipulation.

```python
import pandas as pd

# Convert the Python list into a Pandas DataFrame
df_salaries = pd.DataFrame(salaries, columns=['Salary'])

# Calculate statistics using Pandas built-in methods
pandas_total = df_salaries['Salary'].sum()
pandas_average = df_salaries['Salary'].mean()
pandas_highest = df_salaries['Salary'].max()
pandas_lowest = df_salaries['Salary'].min()

print("--- Pandas Analysis ---")
print(f"Total Salary: ${pandas_total:,}")
print(f"Average Salary: ${pandas_average:,.2f}")
print(f"Highest Salary: ${pandas_highest:,}")
print(f"Lowest Salary: ${pandas_lowest:,}")

print("\n--- Pandas Summary Statistics ---")
# .describe() automatically generates count, mean, std, min, max, and quartiles
print(df_salaries.describe())
```
